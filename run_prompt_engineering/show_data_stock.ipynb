# 輸出一段時間中的新聞資料 tw

```python
import pandas as pd
import json
import os
from datetime import datetime, timedelta

stock_id = '2330'
stock_name = '台積電'
```

```python
def print_news(st, et, stock_id, stock_name):
    # stock_ids = ['2308', '2317', '2330', '2382', '2412', '2454', '2881', '2882', '2891', '6505']
    # stock_names = ['台達電', '鴻海', '台積電', '廣達', '中華電', '聯發科', '富邦金', '國泰金', '中信金', '台塑化']
    
    path_price = f"{os.path.dirname(os.path.abspath(os.getcwd()))}/history_data/tw/stock_price/{stock_id}.csv"

    df_price_his = pd.read_csv(path_price, encoding='utf-8')
    df_price_his['Date'] = pd.to_datetime(df_price_his['Date'], format='%Y%m%d')
    df_price_his = df_price_his[(st <= df_price_his['Date']) & (df_price_his['Date'] <= et)]
    
    news_rise = ''
    news_fall = ''
    for index, row in df_price_his.iterrows():
        news_daily = ""
        path_news_title = os.path.dirname(os.path.abspath(os.getcwd())) + "/history_data/tw/news_title/" + stock_id + "news_title.json"
        
        news_date = row['Date']
        last_day = row['Date'] - timedelta(days=1)
        rtn = (row['Close'] - row['Open']) / row['Open']

        
        with open(path_news_title, 'r', encoding='utf-8') as f:
            news_data = json.load(f)
            news_stock = ''
            for t in news_data[last_day.strftime('%Y%m%d')]:
                headline = t.get("headline")
                # content = t.get("content")
                news_stock += f"{headline}.  "
                # 刪掉t的換行符號
                news_stock = news_stock.replace('\n', '')
                # 刪除股價名稱還有
                # t = t.replace(stock_name, f'company')
                # news_stock += ','
                # 最後一個逗號刪掉
            news_stock = news_stock[:-1]
            news_daily += f"{news_stock}"

        if rtn > 0:
            # news += f"rtn: {round(rtn,5)}, {news_daily}\n"
            news_rise += f"{news_daily}\n"

        if rtn < 0:
            # news += f"rtn: {round(rtn,5)}, {news_daily}\n"
            news_fall += f"{news_daily}\n"

    return news_rise, news_fall

start_day = datetime(2024, 3, 10)
end_day = datetime(2024, 3, 30)

news_rise, news_fall = print_news(start_day, end_day, stock_id, stock_name)

print(
f"""

以下的新聞出現後隔日{stock_name}的股價上漲
『{news_rise}』

以下的新聞出現後{stock_name}的股價下跌
『{news_fall}』

請依據{stock_name}新聞比較兩者的不同
列出10個在上漲會出現的可能原因
再幫我列出10個下跌可能會出現的原因.""")
```
