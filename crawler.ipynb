```python
import time

from selenium import webdriver
from selenium.webdriver.common.by import By

# 找到個股網址 只能用鉅亨
url = 'https://www.cnyes.com/search/news?keyword=%E7%B5%B1%E4%B8%80%E8%B6%85'
driver = webdriver.Chrome()
driver.get(url)
news_url_l = []

for i in range(30):
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(1)

# for i in range(20):
#     print("按下空白鍵", i)
#     actions = ActionChains(driver)
#     actions.send_keys(Keys.SPACE)  # 模擬按下空白鍵
#     actions.perform()
#     time.sleep(1)

element = driver.find_elements(By.XPATH, '//a[contains(@class, "jsx-1986041679") and contains(@class, "news")]')

for e in element:
    # 獲取該元素的 href 屬性
    href = e.get_attribute('href')
    news_url_l.append(href)
    print("目標元素的網址:", href)

driver.quit()
```

```python
news_url_l2 = news_url_l[267:]
```

```python
import re
from datetime import datetime
#顯性等待​
from selenium.webdriver.support.ui import WebDriverWait
#預期條件函數，網頁元素在瀏覽器中的的出現、可見、可點擊
from selenium.webdriver.support import expected_conditions
#等待超時異常處理
from selenium.common.exceptions import TimeoutException, NoSuchElementException

stock_id = "1216"

p = ""
time0 = ""
for index, link in enumerate(news_url_l2):
    print(f'[{index+1}/{len(news_url_l2)}], link:{link}')
    
    driver = webdriver.Chrome()
    driver.get(link)
    driver.implicitly_wait(20) # seconds
    try:
        # time
        time_element = driver.find_element(By.XPATH,"//p[@class='a1wfawlv']")
        time_text = time_element.text
        match = re.search(r'\d{4}-\d{2}-\d{2} \d{2}:\d{2}', time_text)
        if match:
            time_string = match.group()
            time_datetime = datetime.strptime(time_string, '%Y-%m-%d %H:%M')
            formatted_time = time_datetime.strftime('%Y%m%d')
        
        if time0 != formatted_time:
            p = ""
            time0 = formatted_time
        
        #創造一個顯性等待，等待時間20秒，每1秒檢查一次、忽略找不到元素的錯誤​
        # driver_wait = WebDriverWait(driver, timeout=20, poll_frequency=1)
        
        # 文章內容
        paragraphs = driver.find_elements(By.XPATH,  "//main[@id='article-container']//p")
        
        # 提取每個 <p> 標籤的文本
        for paragraph in paragraphs:
            text = paragraph.text
            p = p + text
        
        # print("link" + link)
        print("時間：" + formatted_time)
        # print("段落的資料：", p)
        
        filename = "stock_news/"+stock_id+"/"+stock_id+"_news/"+stock_id+"n_" + formatted_time
        with open(filename, "w") as f:
            f.write(p)
            
    except Exception as e:
        print(e)
        pass
    
    driver.quit()
```
