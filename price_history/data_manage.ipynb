```python
import yfinance as yf
import os

folder_list = os.listdir(os.path.dirname(os.path.abspath(os.getcwd())) + "/stock_news")
folder_list = [folder_list for folder_list in folder_list if folder_list != '.DS_Store']
print(folder_list)

def fetch_stock_data(ticker_symbol, start_date, end_date):
    print("Fetching data for " + ticker_symbol + " from " + start_date + " to " + end_date)
    ticker_data = yf.Ticker(ticker_symbol)
    return ticker_data.history(period='1d', start=start_date, end=end_date)

id = "1216"
data = fetch_stock_data(id + ".tw", "2022-01-01", "2023-12-31") # 櫃買需換成two
# print(data)
data.index = data.index.strftime('%Y%m%d')
data.to_csv(id + 'price.csv')

# # 使用範例
# for id in folder_list:
#     try:
#         data = fetch_stock_data(id + ".two", "2022-01-01", "2023-12-31") # 櫃買需換成two
#         # print(data)
#         data.index = data.index.strftime('%Y%m%d')
#         data.to_csv(id + 'price.csv')
#     except Exception as e:
#         print("Error: " + id + " " + str(e))
#         pass
```
